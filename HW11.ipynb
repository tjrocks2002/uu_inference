## 7.

```python
import pandas as pd
import numpy as np
df = pd.read_csv('./data/ames_prices.csv')
df['dum'] = df['Central.Air']=='Y'
```

### A)

```python
caprop = df['dum'].mean()
print(f'Approximately {caprop*100:.2f}% of homes have Central Air')
```

### B)

```python
myseed = 1231

np.random.seed(myseed)
R = 10_000
N = len(df['dum'])
results = []

for t in range (R):
    indx = np.random.randint(0, N, N)
    results.append(df['dum'].iloc[indx].mean())

print(f'Seed: {myseed}\nReplications: {R}')
```

### C)

```python
se = np.std(results, ddof = 1)
print(f'Standard Error: {se:.5f}')
```

### D)

```python
cv90 = 1.645
cv95 = 1.960
cv99 = 2.576

#90%
min90 = caprop - se*cv90
max90 = caprop + se*cv90

#95%
min95 = caprop - se*cv95
max95 = caprop + se*cv95

#99%
min99 = caprop - se*cv99
max99 = caprop + se*cv99

print(f'90% CI: ({min90:.5f},{max90:.5f})\n95% CI: ({min95:.5f},{max95:.5f})\n99% CI: ({min99:.5f},{max99:.5f})')
```

### E)

90% CI (alpha = 0.10): (0.92545,0.94076) -> Reject H0\
95% CI (alpha = 0.05): (0.92398,0.94223) -> Reject H0\
99% CI (alpha = 0.01): (0.92111,0.94510) -> Reject H0

At significance levels of .10, .05, and .01 we reject the null hypothesis that p = 0.90. This means that with 90%, 95%, and 99% confidence, we reject the assumption that the true proportion of homes with central air is 0.90.

## 8.

### A)

```python
comparison = pd.DataFrame({
    'asinh_area':  np.arcsinh(df['area']),
    'asinh_price': np.arcsinh(df['price'])
})

ogc = comparison.corr().iloc[0,1]

np.random.seed(myseed)
R = 10_000
N = len(comparison['asinh_area'])
results2 = []

for t in range(R):
    indx = np.random.randint(0, N, N)
    resample = comparison.iloc[indx]
    results2.append(resample.corr().iloc[0, 1])

print(f'Pearson Correlation: {ogc:.4f}\nSeed: {myseed}\nReplications: {R}')
```

### B)

```python
se2 = np.std(results2, ddof = 1)
print(f'Standard Error: {se2:.5f}')
```

### C)

```python
cv90 = 1.645
cv95 = 1.960
cv99 = 2.576

#90%
min90 = ogc - se2*cv90
max90 = ogc + se2*cv90

#95%
min95 = ogc - se2*cv95
max95 = ogc + se2*cv95

#99%
min99 = ogc - se2*cv99
max99 = ogc + se2*cv99

print(f'90% CI: ({min90:.5f},{max90:.5f})\n95% CI: ({min95:.5f},{max95:.5f})\n99% CI: ({min99:.5f},{max99:.5f})')
```

### D)

90% CI: (0.70656,0.73978) -> Reject H0 at alpha = 0.10\
95% CI: (0.70338,0.74296) -> Reject H0 at alpha = 0.05\
99% CI: (0.69716,0.74918) -> Reject H0 at alpha - 0.01

At all 3 significance levels, we reject H0: rho = 0. This means that we reject the assumption that there is no correlation between the arcsinh of area and the arcsinh of price, noting that 0 falls below the confidence intervals.

### E)

Area and price need to remain paired during resampling because each record has a unique area-price combo. Separating them would remove the relationship between the two that we are attempting to observe.

### D)

### A)

```python
df = pd.read_csv('./data/nhanes_data_17_18.csv')
df = df[(df['AgeInYearsAtScreening'] >= 18) & df['SleepHoursWeekdaysOrWorkdays'].notna()]

wksleep = df['SleepHoursWeekdaysOrWorkdays'].mean()

np.random.seed(myseed)
R = 10_000
N = len(df['SleepHoursWeekdaysOrWorkdays'])
results4 = []

for t in range (R):
    indx = np.random.randint(0, N, N)
    results4.append(df['SleepHoursWeekdaysOrWorkdays'].iloc[indx].mean())

print(f'Mean Weekday Sleep Time: {wksleep:.2f}\nSeed: {myseed}\nReplications: {R}')
```

### B)

```python
se4 = np.std(results4, ddof = 1)
print(f'Standard Error: {se4:.5f}')
```

### C)

```python
cv90 = 1.645
cv95 = 1.960
cv99 = 2.576

#90%
min90 = wksleep - se4*cv90
max90 = wksleep + se4*cv90

#95%
min95 = wksleep - se4*cv95
max95 = wksleep + se4*cv95

#99%
min99 = wksleep - se4*cv99
max99 = wksleep + se4*cv99

print(f'90% CI: ({min90:.5f},{max90:.5f})\n95% CI: ({min95:.5f},{max95:.5f})\n99% CI: ({min99:.5f},{max99:.5f})')
```

### D)

90% CI: (7.58454,7.65795) -> Reject H0 at alpha = 0.10\
95% CI: (7.57751,7.66498) -> Reject H0 at alpha = 0.05\
99% CI: (7.56376,7.67872) -> Reject H0 at alpha = 0.01

### E)

At all 3 significance levels we rejected the null hypothesis that the true mean weekday sleep time was 8 hours. Statistically this is significant, but practically we are looking at a difference in, at alpha 0.10 for example, 0.343 hours or 20.58 minutes. This is not a practically significant amount of sleep to lose or gain, so practically speaking, while 8 hours is not statistically significant, 8 hours is a practically useful estimate of the true mean.
